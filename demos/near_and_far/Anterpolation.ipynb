```python
import numpy as np
import numpy.linalg as la
import matplotlib.pyplot as plt
```

Some Chebyshev utilities up front:

```python
def Tk(k, x):
    return np.cos(k*np.arccos(x))

nnodes = 20
k = nnodes-1
nodes = np.cos(np.arange(k+1)/k*np.pi)

xplot = np.linspace(-1, 1, 1000)
plt.plot(xplot, Tk(k, xplot))
plt.plot(nodes, Tk(k, nodes), "x")
```

Now construct an interpolation toolkit with `vdm(targets)` and `interp_to_targets(targets)`, and test it:

```python
#clear
def vdm(targets):
    return Tk(np.arange(k+1), targets[:, None])
        
def interp_to_targets(targets, nodes):
    return vdm(targets) @ la.inv(vdm(nodes))

print(la.norm(np.sin(5*xplot) - interp_to_targets(xplot, nodes)@np.sin(5*nodes), np.inf))
```

```python
N = 1000
sources = np.random.rand(N) * 2 - 1
charges = np.random.randn(N)

targets = np.random.rand(N)*2 - 1 - 4

def interaction_mat(targets, sources):
    return np.log(np.abs(targets[:, None] - sources))
```

Try out target-space interpolation as a matrix identity:

```python
tgt_nodes = nodes - 4
la.norm(
    interp_to_targets(targets+4, tgt_nodes+4) @ interaction_mat(tgt_nodes, sources)
    -
    interaction_mat(targets, sources),
    2
)
```

Now try out source interpolation ("anterpolation") as a matrix identity:

```python
#clear
la.norm(
    interaction_mat(targets, nodes) @ interp_to_targets(sources, nodes).T
    -
    interaction_mat(targets, sources), 
    2)
```
